```python
def count_up_to(n):
    i = 1
    while i <= n:
       yield i
       i += 1

gen_func = count_up_to(5)
for number in gen_func:
    # if (number == 2):
    #     gen_func.close()
        print(number)


```

```python
def sum_digit(n):
   if n == 0:
     return 0
   else:
      return (n % 10) + sum_digit(n // 10)
   
print(sum_digit(133))
```

```python
class students:
    def __init__(self,  name, marks=[]):
        self.name = name
        self.marks = marks

    def average(self):
        
        self.average_marks = sum(self.marks) / len(self.marks)
        return self.average_marks
    def grade(self):
        if self.average_marks >= 80:
            return "A"
        if self.average_marks >= 70:
            return "B"
        if self.average_marks >= 60:
            return "C"
        if self.average_marks >= 50:
            return "D"
        else: "F"

abdulsalam = students("abdulsalam",[80, 90, 70, 80])

print(abdulsalam.name, abdulsalam.average(), abdulsalam.grade())





        

```

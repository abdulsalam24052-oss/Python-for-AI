```python
class students:
    def __init__(self,  name, marks=[]):
        self.name = name
        self.marks = marks

    def average(self):
        
        self.average_marks = sum(self.marks) / len(self.marks)
        return self.average_marks
    def grade(self):
        if self.average_marks >= 80:
            return "A"
        if self.average_marks >= 70:
            return "B"
        if self.average_marks >= 60:
            return "C"
        if self.average_marks >= 50:
            return "D"
        else: "F"

abdulsalam = students("abdulsalam",[80, 90, 70, 80])

print(abdulsalam.name, abdulsalam.average(), abdulsalam.grade())





        

```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"   # parent's version


class Dog(Animal):
    def speak(self):
        base = super().speak()      # run Animal.speak(), store its return value
        return base + " — Woof!"   # extend it; don't replace it


class GuardDog(Dog):
    def speak(self):
        base = super().speak()      # runs Dog.speak() (which already called Animal's)
        return base + " BARK BARK!"


print(Animal("Thing").speak())
print(Dog("Tommy").speak())
print(GuardDog("Rex").speak())
```

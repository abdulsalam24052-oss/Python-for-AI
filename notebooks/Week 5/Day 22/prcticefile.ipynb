```python
import torch
from torch import nn
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
students = pd.read_csv('students_100.csv')

X = torch.tensor(
    students[["study_hours", "phone_hours"]].values,
    dtype = torch.float32,
)

Y = torch.tensor(
    students['maths'].values,
    dtype = torch.float32,
)

print(X.shape, Y.shape)




```

```python
torch.manual_seed(42)
model = nn.Sequential(
    nn.Linear(2,3), #input layer(2 elements) #hidden layer(3 elements) #1 output #6 weights bcz 2x3
    nn.ReLU(),
    nn.Linear(3,4),
    nn.ReLU(),
    nn.Linear(4,1),
)

loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)

```

```python
losses = []
for epoch in range (300):
    guess = model(X)
    loss = loss_fn(guess, Y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    losses.append(loss.item())
    if epoch % 50 == 0:
        print(f'epoch {epoch:3}: loss={loss.item():1f}')
        
        
```

```python
plt.plot(losses)
plt.ylim(0, 1000)
plt.title("Loss during training")
plt.xlabel("Epoch")
plt.ylabel("Loss (MSE)")
plt.show()
```

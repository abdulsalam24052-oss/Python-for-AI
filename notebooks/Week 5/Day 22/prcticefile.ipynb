```python
import torch
from torch import nn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
students =  pd.read_csv("students_100.csv")
print(students.shape)
X = torch.tensor(
    students[["study_hours", "phone_hours"]].values, 
    dtype = torch.float32,)
print(X.shape)
Y = torch.tensor(
    students[["maths"]].values,
    dtype=torch.float32
    )
print(Y.shape)




```

```python
torch.manual_seed(42)
model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Linear(8, 1),
)


loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.1)

print("Weignts + biases to learn:", sum(p.numel() for p in model.parameters()))

```

```python
losses = []
for epoch in range (300):
    guess = model(X)
    loss = loss_fn(guess, Y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    losses.append(loss.item())
    if epoch % 50 == 0:
        print(f'epoch {epoch:3}: loss={loss.item():1f}')
        
        
```

```python
plt.plot(losses)
plt.ylim(0, 1000)
plt.title("Loss during training")
plt.xlabel("Epoch")
plt.ylabel("Loss (MSE)")
plt.show()
```
